### **EL SIGUIENTE EJEMPLO ES SIN UTILIZAR MACHINE LEARNING**

Para el siguiente ejemplo:
    *Observación humana:
    Hasta 3 horas → 0
    Desde 4 horas → 1
    Entonces la frontera de decisión es:*

    Clasificación : asignar etiqueta.
    Existe una regla de decisión.
    La clasificación separa en categorías.
    Todavía NO hay aprendizaje automático.

```python

#todo: Frontera
#! si horas >= 4 → clase 1
#? si horas < 4 → clase 0

horas_estudio = [1, 2, 3, 4, 5, 6]
resultado_real = [0, 0, 0, 1, 1, 1]

for h in horas_estudio:
    
    if h >= 4:
        prediccion = 1
    else:
        prediccion = 0
        
    print("Horas:", h, "→ Predicción:", prediccion)
```

### **EL SIGUIENTE EJEMPLO SI UTILIZA EL MACHINE LEARNING**
*1.-Primero hicimos una regla manual.
2.-Luego dejamos que la computadora ajuste la regla.
3.- Esa regla se define con peso y bias.
4.- Cuando se equivoca → corrige.
5.- Eso es aprendizaje.*

```python
horas_estudio = [1, 2, 3, 4, 5, 6]
resultado_real = [0, 0, 0, 1, 1, 1]

peso = 0
bias = 0
alpha = 0.1

for i in range(len(horas_estudio)):
    
    x = horas_estudio[i]
    y_real = resultado_real[i]
    
    z = peso * x + bias
    
    if z >= 0:
        y_pred = 1
    else:
        y_pred = 0
        
    error = y_real - y_pred
    
    peso = peso + alpha * error * x
    bias = bias + alpha * error

print("Peso:", peso)
print("Bias:", bias)
```
